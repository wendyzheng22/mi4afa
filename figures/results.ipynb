```python
import json
import numpy as np

models = ["qwen0.5", "qwen1.5", "qwen7", "llama1", "llama3", "gemma1", "gemma4", "gpt-5.6-sol", "gpt-5.6-terra", "claude-sonnet-5", "claude-opus-4-8", "claude-opus-5"]
datasets = ["algo", "craft"]
res = {d: {mo: {} for mo in models} for d in datasets}

prompt_res = json.load(open("./../prompting/results/summary.json"))
for key in prompt_res:
    model, data = key.split("_")[0], key.split("_")[1]
    if "with_step" not in key:
        continue
    if data not in datasets:
        continue
    res[data][model] = {k: prompt_res[key][k] for k in ["agent_acc", "step_acc"]}

prompt_res = json.load(open("./../prompting_big/results/summary.json"))
for key in prompt_res:
    model, data = key.split("_")[0], key.split("_")[1]
    if "with_step" not in key:
        continue
    if data not in datasets:
        continue
    res[data][model] = {k: prompt_res[key][k] for k in ["agent_acc", "step_acc"]}


for mo in models:
    line = ""
    for d in datasets:
        if "agent_acc" not in res[d][mo]:
            line += "n/a\tn/a\t"
        else:
            line += f"{np.round(res[d][mo]['agent_acc'], 4) * 100:.2f}\t"
            line += f"{np.round(res[d][mo]['step_acc'], 4) * 100:.2f}\t"
    print(line)
```
